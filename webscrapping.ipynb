```python
items = []
for article in articles:
    title = article.find("h3").find("a")["title"]
    price = article.select_one("p.price_color").text.split("£")[1]
    rating_element =article.select_one("p.star-rating")
    rating = rating_element["class"][1]
    items.append([title, price, rating])
print(items)
```

```python
import pandas as pd
df=pd.DataFrame(items, columns=["Book Title", "Price","Rating"])
print(df.head())
```

```python
from bs4 import BeautifulSoup
import pandas as pd
items = []
for i in range (1,51):
    with open(f"htmls/page{i}.html", "r",encoding="utf-8")as f:
        content = f.read()
    soup = BeautifulSoup(content, "html.parser")
    articles = soup.select("article.product_pod")
    for article in articles:
        title = article.find("h3").find("a")["title"]
        price = article.select_one("p.price_color").text.split("£")[1]
        rating = rating_element["class"][1]
        availability = article.select_one("p.instock.availability").text.strip()
        items.append([title, price, rating, availability])
    print(f"parsed page {i} -{len(items)}")
df=pd.DataFrame(items, columns=["Book Title", "Price","Rating", "Availability"])
df.to_csv("books_data.csv",index=False)
print(f"\nTotal books scraped: {len(df)}")
print(df.head())
```
